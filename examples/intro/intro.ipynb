```python
import shape2graph
import ipywidgets as widgets
```

```python
m = shape2graph.Map(height="800px")
m.add_basemap(basemap="Esri.NatGeoWorldMap")
m
```

```python
m.add_basemap_gui(position="topright")
```

```python
zoom_slider = widgets.IntSlider(value=10, min=0, max=24, description="Zoom:")

widgets.jslink((zoom_slider, "value"), (m, "zoom"))
m.add_widget(zoom_slider, position="topright")
```
